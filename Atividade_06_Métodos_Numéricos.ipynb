```python
import time
import math

def bissecao(f, a, b, tol, max_iter):
    if f(a) * f(b) >= 0:
        return None, 0, True, 0.0
    start = time.perf_counter()
    for i in range(max_iter):
        c = (a + b) / 2
        if abs(f(c)) < tol or (b - a) / 2 < tol:
            return c, i + 1, False, time.perf_counter() - start
        if f(a) * f(c) < 0:
            b = c
        else:
            a = c
    return (a + b) / 2, max_iter, True, time.perf_counter() - start

def falsa_posicao(f, a, b, tol, max_iter):
    if f(a) * f(b) >= 0:
        return None, 0, True, 0.0
    start = time.perf_counter()
    c = a
    for i in range(max_iter):
        c_old = c
        fa, fb = f(a), f(b)
        if fb - fa == 0:
            return c, i + 1, True, time.perf_counter() - start
        c = (a * fb - b * fa) / (fb - fa)
        if abs(f(c)) < tol or abs(c - c_old) < tol:
            return c, i + 1, False, time.perf_counter() - start
        if fa * f(c) < 0:
            b = c
        else:
            a = c
    return c, max_iter, True, time.perf_counter() - start

def ponto_fixo(g, x0, tol, max_iter):
    start = time.perf_counter()
    x = x0
    for i in range(max_iter):
        try:
            x_new = g(x)
            if abs(x_new - x) < tol:
                return x_new, i + 1, False, time.perf_counter() - start
            x = x_new
        except:
            return x, i + 1, True, time.perf_counter() - start
    return x, max_iter, True, time.perf_counter() - start

def newton_raphson(f, df, x0, tol, max_iter):
    start = time.perf_counter()
    x = x0
    for i in range(max_iter):
        try:
            deriv = df(x)
            if deriv == 0:
                return x, i + 1, True, time.perf_counter() - start
            x_new = x - f(x) / deriv
            if abs(x_new - x) < tol:
                return x_new, i + 1, False, time.perf_counter() - start
            x = x_new
        except:
            return x, i + 1, True, time.perf_counter() - start
    return x, max_iter, True, time.perf_counter() - start

def secante(f, x0, x1, tol, max_iter):
    start = time.perf_counter()
    for i in range(max_iter):
        try:
            f0, f1 = f(x0), f(x1)
            if f1 - f0 == 0:
                return x1, i + 1, True, time.perf_counter() - start
            x_new = x1 - f1 * (x1 - x0) / (f1 - f0)
            if abs(x_new - x1) < tol:
                return x_new, i + 1, False, time.perf_counter() - start
            x0, x1 = x1, x_new
        except:
            return x1, i + 1, True, time.perf_counter() - start
    return x1, max_iter, True, time.perf_counter() - start

equacoes = [
    (
        "a",
        lambda x: 2*x**4 + 4*x**3 + 3*x**2 - 10*x - 15,
        lambda x: (2*x**4 + 4*x**3 + 3*x**2 - 15) / 10,
        lambda x: 8*x**3 + 12*x**2 + 6*x - 10,
        0, 3
    ),
    (
        "b",
        lambda x: (x + 3)*(x + 1)*(x - 2)**3,
        lambda x: x - ((x + 3)*(x + 1)*(x - 2)**3)/100,
        lambda x: (x + 1)*(x - 2)**3 + (x + 3)*(x - 2)**3 + 3*(x + 3)*(x + 1)*(x - 2)**2,
        0, 5
    ),
    (
        "c",
        lambda x: 5*x**3 + x**2 - math.exp(1 - 2*x) + math.cos(x) + 20,
        lambda x: x - (5*x**3 + x**2 - math.exp(1 - 2*x) + math.cos(x) + 20)/100,
        lambda x: 15*x**2 + 2*x + 2*math.exp(1 - 2*x) - math.sin(x),
        -5, 5
    ),
    (
        "d",
        lambda x: math.sin(x)*x + 4,
        lambda x: x - (math.sin(x)*x + 4)/10,
        lambda x: math.sin(x) + x*math.cos(x),
        1, 5
    ),
    (
        "e",
        lambda x: ((x - 3)**5) * math.log(x) if x > 0 else float('inf'),
        lambda x: x - (((x - 3)**5) * math.log(x))/100 if x > 0 else float('inf'),
        lambda x: 5*((x - 3)**4)*math.log(x) + ((x - 3)**5)/x if x > 0 else float('inf'),
        2, 5
    ),
    (
        "f",
        lambda x: x**10 - 1,
        lambda x: x - (x**10 - 1)/10,
        lambda x: 10*x**9,
        0.8, 1.2
    )
]

tol = 1e-10
max_iter = 200

for nome, f, g, df, a, b in equacoes:
    print(f"Equação ({nome})")

    raiz, it, erro, tempo = bissecao(f, a, b, tol, max_iter)
    rz = f"{raiz:.10f}" if raiz is not None else "N/A"
    print(f"Bisseção      | Raiz: {rz:<15} | Iter: {it:<3} | Erro: {str(erro):<5} | Tempo: {tempo:.6f}s")

    raiz, it, erro, tempo = falsa_posicao(f, a, b, tol, max_iter)
    rz = f"{raiz:.10f}" if raiz is not None else "N/A"
    print(f"Falsa Posição | Raiz: {rz:<15} | Iter: {it:<3} | Erro: {str(erro):<5} | Tempo: {tempo:.6f}s")

    raiz, it, erro, tempo = ponto_fixo(g, b, tol, max_iter)
    rz = f"{raiz:.10f}" if isinstance(raiz, (int, float)) and not math.isnan(raiz) and not math.isinf(raiz) else "N/A"
    print(f"Ponto Fixo    | Raiz: {rz:<15} | Iter: {it:<3} | Erro: {str(erro):<5} | Tempo: {tempo:.6f}s")

    raiz, it, erro, tempo = newton_raphson(f, df, b, tol, max_iter)
    rz = f"{raiz:.10f}" if isinstance(raiz, (int, float)) and not math.isnan(raiz) and not math.isinf(raiz) else "N/A"
    print(f"Newton-Raphson| Raiz: {rz:<15} | Iter: {it:<3} | Erro: {str(erro):<5} | Tempo: {tempo:.6f}s")

    raiz, it, erro, tempo = secante(f, a, b, tol, max_iter)
    rz = f"{raiz:.10f}" if isinstance(raiz, (int, float)) and not math.isnan(raiz) and not math.isinf(raiz) else "N/A"
    print(f"Secante       | Raiz: {rz:<15} | Iter: {it:<3} | Erro: {str(erro):<5} | Tempo: {tempo:.6f}s\n")
```
