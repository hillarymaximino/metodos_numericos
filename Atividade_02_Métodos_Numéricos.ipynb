```python
import numpy as np
np.set_printoptions(precision = 8, suppress = True)

num_termos = int(input("Digite o numero de termos: "))
soma_acumulada = 0

for i in range (num_termos):
  numerador = (-1) ** i
  denominador = 2 * i + 1
  termo = 4* (numerador / denominador)
  soma_acumulada = soma_acumulada + termo
  print(soma_acumulada)

```
