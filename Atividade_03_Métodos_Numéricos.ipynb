**Exercício 1.1**

```python
import numpy as np

a = int(input("Digite o valor de a: "))
x = 2.0
valor_real = a ** 0.5
print(f"Raiz quadrada de {a}: {valor_real}\n")

for i in range(1, 11):
    x = (x + a / x) / 2
    erro_relativo = abs((valor_real - x) / valor_real) * 100
    print(f"Interação {i}")
    print(f"Aproximação: {x}")
    print(f"Erro relativo: {erro_relativo:.2f}%\n")
```

**Exercício 1.4**

```python
i = 0.1
x = np.exp(1/i)/(1+np.exp(1/i))
print(f"Valor de x: {x}")
y = 1/(np.exp(-1/i)+1)
print(f"Valor de y: {y}\n")

i = 0.01
x = np.exp(1/i)/(1+np.exp(1/i))
print(f"Valor de x: {x}")
y = 1/(np.exp(-1/i)+1)
print(f"Valor de y: {y}\n")

i = 0.001
x = np.exp(1/i)/(1+np.exp(1/i))
print(f"Valor de x: {x}")
y = 1/(np.exp(-1/i)+1)
print(f"Valor de y: {y}\n")
```

A segunda expressão é mais adequada, pois no exemplo em que i = 0.001 a exp(1/0.001) causa um overflow, o que torna a expressão inf/1+inf.

Já na segunda expressão para um i pequeno exp(-1000) simplesmente vira zero (underflow)

**Exercício 1.6**

```python
import numpy as np
import matplotlib.pyplot as plt

def arctan_taylor(x, n_termos):
    soma = 0
    for n in range(n_termos):
        termo = ((-1)**n * x**(2*n + 1)) / (2*n + 1)
        soma += termo
    return soma

termos = np.arange(1, 21)
pi_real = np.pi
metodos = {
    "Machin": lambda n: 4 * (4 * arctan_taylor(1/5, n) - arctan_taylor(1/239, n)),
    "Hutton": lambda n: 4 * (arctan_taylor(1/2, n) + arctan_taylor(1/3, n)),
    "Clausen": lambda n: 4 * (2 * arctan_taylor(1/3, n) + arctan_taylor(1/7, n)),
    "Dase": lambda n: 4 * (arctan_taylor(1/2, n) + arctan_taylor(1/5, n) + arctan_taylor(1/8, n))
}

# 3. Cálculo dos erros
plt.figure(figsize=(10, 6))

for nome, formula in metodos.items():
    erros = []
    for n in termos:
        pi_aprox = formula(n)
        erro_abs = np.abs(pi_real - pi_aprox)
        erros.append(erro_abs)

    plt.plot(termos, erros, marker='o', label=nome)

plt.yscale('log')
plt.title('Comparação da Convergência para o Cálculo de $\pi$ (Exercício 1.6)')
plt.xlabel('Número de Termos na Série (N)')
plt.ylabel('Erro Absoluto ($| \pi - \pi_{aprox} |$)')
plt.xticks(termos)
plt.grid(True, which="both", ls="-", alpha=0.5)
plt.legend()
plt.show()
```
