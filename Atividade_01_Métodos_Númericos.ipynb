**Questão 02**

Vetorização: É o conceito de realizar operações matemáticas de forma otimizada sem a necessidade de usar um for ou while, é possível fazer a operação com todos os valores ao mesmo tempo.

Broadcast: É um conjunto de regras que permite ao Numpy realizar operações aritmeticas com arrays de diferentes tamanhos (formas), obedecendo algumas regras de dimensão ele consegue realizar operações mesmo que esses arrays sejam diferentes

**Questão 07**

*a) A + B*

```python
A = np.arange(10,49)
B = np.arange(10,39)

X = A + B
X
```

*(b) A * B*

```python
A = np.arange(10,49)
B = np.arange(10,39)

X = A * B
X
```

*(c) A + C*

```python
A = np.arange(10,49)
C = np.arange(10,20)

X = A + C
X
```

*(d) B · C*

```python
B = np.arange(10,39)
C = np.arange(10,20)

X = B * C
X
```

Explicação a) e c): A e B | A + C possui tamanhos diferentes e como nenhum deles é 1, não é possível realizar broadcast para finalizar a soma


Explicação b) e d): A multipplicação exige que os valores tenham mesmo tamanho (compatível com broadcast), como não é o caso de nenhum deles
