**Exercício 1.2**

```python
A = np.array([[3, 4, -5, 1], [0, 1, 1, -2], [0, 0, 4, -5], [0, 0, 0, 2]], dtype=float)
b = np.array([-10, -1, 3, 2], dtype=float)

n = len(b)
x = np.empty(n)

x[-1] = b[-1] / A[-1, -1]
print(f"Passo 1 (i = {n-1}): x[{n-1}] = {b[-1]} / {A[-1, -1]} = {x[-1]}")

for i in range(n - 2, -1, -1):
    soma_produtos = np.sum(A[i, i + 1 :] * x[i + 1 :])
    x[i] = (b[i] - soma_produtos) / A[i, i]
    print(
        f"Passo {n-i} (i = {i}): x[{i}] = ({b[i]} - {soma_produtos}) / {A[i, i]} = {x[i]}"
    )

print("\nVetor Solução Final x:", x)
```

**Exercício 1.4**

```python
def sist_lin_diagonal(A, b):
    return b / np.diagonal(A)


def sist_lin_tri_sup(A, b):
    n = len(b)
    x = np.empty(n, dtype=float)
    x[-1] = b[-1] / A[-1, -1]
    for i in range(n - 2, -1, -1):
        x[i] = (b[i] - np.sum(A[i, i + 1 :] * x[i + 1 :])) / A[i, i]
    return x


def sist_lin_tri_inf(A, b):
    n = len(b)
    x = np.empty(n, dtype=float)
    x[0] = b[0] / A[0, 0]
    for i in range(1, n):
        x[i] = (b[i] - np.sum(A[i, :i] * x[:i])) / A[i, i]
    return x


def resolver_sistema_trivial(A, b):
    """Identifica o tipo de matriz trivial (Diagonal, Triangular Superior ou Inferior)

    e resolve o sistema Ax = b.
    """
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)

    if A.shape[0] != A.shape[1]:
        raise ValueError("A matriz de coeficientes A deve ser quadrada.")

    if np.allclose(A, np.diag(np.diagonal(A))):
        print("Estrutura identificada: Matriz Diagonal")
        return sist_lin_diagonal(A, b)

    elif np.allclose(A, np.triu(A)):
        print("Estrutura identificada: Matriz Triangular Superior")
        return sist_lin_tri_sup(A, b)

    elif np.allclose(A, np.tril(A)):
        print("Estrutura identificada: Matriz Triangular Inferior")
        return sist_lin_tri_inf(A, b)

    else:
        raise ValueError(
            "A matriz fornecida não se enquadra nos casos triviais (diagonal, triangular superior ou inferior)."
        )

A_diag = np.diag([1, 4, 2])
b_diag = np.array([4, 2, 2])
x_diag = resolver_sistema_trivial(A_diag, b_diag)
print("Solução Diagonal:", x_diag, "\n")

A_sup = np.array([[3, 4, -5, 1], [0, 1, 1, -2], [0, 0, 4, -5], [0, 0, 0, 2]])
b_sup = np.array([-10, -1, 3, 2])
x_sup = resolver_sistema_trivial(A_sup, b_sup)
print("Solução Triangular Superior:", x_sup, "\n")

A_inf = np.array([[3, 0, 0, 0], [2, 1, 0, 0], [1, 0, 1, 0], [1, 1, 1, 1]])
b_inf = np.array([4, 2, 4, 2])
x_inf = resolver_sistema_trivial(A_inf, b_inf)
print("Solução Triangular Inferior:", x_inf)
```
