**Exercise 2-3**

```python
import numpy as np

def CRIAR_VETOR_UNITARIO(VETOR_ENTRADA):
    NORMA_VETOR = np.linalg.norm(VETOR_ENTRADA)
    if NORMA_VETOR == 0:
        return VETOR_ENTRADA
    return VETOR_ENTRADA / NORMA_VETOR

VETOR_TESTE = np.array([3, 4])
print("VETOR UNITÁRIO:", CRIAR_VETOR_UNITARIO(VETOR_TESTE))
```

**Exercise 2-4**

```python
import numpy as np

def CRIAR_VETOR_UNITARIO(VETOR_ENTRADA):
    NORMA_VETOR = np.linalg.norm(VETOR_ENTRADA)
    if NORMA_VETOR == 0:
        return VETOR_ENTRADA
    return VETOR_ENTRADA / NORMA_VETOR

def ALTERAR_MAGNITUDE(VETOR_ENTRADA, MAGNITUDE_DESEJADA):
    VETOR_UNITARIO = CRIAR_VETOR_UNITARIO(VETOR_ENTRADA)
    VETOR_FINAL = VETOR_UNITARIO * MAGNITUDE_DESEJADA
    return VETOR_FINAL

VETOR_TESTE = np.array([3, 4])
print("VETOR COM MAGNITUDE 10:", ALTERAR_MAGNITUDE(VETOR_TESTE, 10))
```

**Exercise 2-9**

```python
import numpy as np

VETOR_ALVO_T = np.random.randn(3)
VETOR_REFERENCIA_R = np.random.randn(3)

ESCALAR_PROJECAO = np.dot(VETOR_ALVO_T, VETOR_REFERENCIA_R) / np.dot(VETOR_REFERENCIA_R, VETOR_REFERENCIA_R)
COMPONENTE_PARALELA = ESCALAR_PROJECAO * VETOR_REFERENCIA_R
COMPONENTE_PERPENDICULAR = VETOR_ALVO_T - COMPONENTE_PARALELA

SOMA_COMPONENTES = COMPONENTE_PARALELA + COMPONENTE_PERPENDICULAR
PRODUTO_ORTOGONALIDADE = np.dot(COMPONENTE_PERPENDICULAR, VETOR_REFERENCIA_R)

print("SOMA DAS COMPONENTES (IGUAL AO ALVO):", SOMA_COMPONENTES)
print("PRODUTO ESCALAR (ORTOGONALIDADE):", PRODUTO_ORTOGONALIDADE)
```

```python
import numpy as np

LISTA_ESCALARES = [1, 2, -3]
LISTA_VETORES = [np.array([4, 5, 1]), np.array([-4, 0, -4]), np.array([1, 3, 2])]

DIMENSAO = len(LISTA_VETORES[0])
VETOR_RESULTADO = np.zeros(DIMENSAO)

for INDICE in range(len(LISTA_ESCALARES)):
    VETOR_RESULTADO += LISTA_ESCALARES[INDICE] * LISTA_VETORES[INDICE]

print("VETOR RESULTANTE:", VETOR_RESULTADO)
```
